```python
import requests
from bs4 import BeautifulSoup

# Function to scrape Books to Scrape
def scrape_books_to_scrape():
    url = "http://books.toscrape.com/"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    products = soup.select(".product_pod")

    for product in products[:5]:  # Limit to first 5 for demo
        name = product.h3.a["title"]
        price = product.select_one(".price_color").text
        stock = product.select_one(".instock.availability").text.strip()
        website = "Books to Scrape"

        print(f"Website: {website}")
        print(f"Product: {name}")
        print(f"Price: {price}")
        print(f"Stock Status: {stock}")
        print("-" * 40)

# Function to scrape Webscraper.io demo e-commerce site
def scrape_webscraper_demo():
    url = "https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    products = soup.select(".thumbnail")

    for product in products[:5]:
        name = product.select_one(".title").text.strip()
        price = product.select_one(".price").text.strip()
        stock = "In Stock"  # Simulated; this site doesn’t show actual stock
        website = "Webscraper Demo"

        print(f"Website: {website}")
        print(f"Product: {name}")
        print(f"Price: {price}")
        print(f"Stock Status: {stock}")
        print("-" * 40)

# Run both scrapers
scrape_books_to_scrape()
scrape_webscraper_demo()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

# Sample data for wireless earbuds
data = [
    {
        "product_name": "Sony WF-1000XM4",
        "website": "Amazon.sg",
        "current_price": 228,
        "original_price": 299,
        "stock_status": "In Stock",
        "scraping_date": datetime.today().strftime('%Y-%m-%d')
    },
    {
        "product_name": "AirPods Pro",
        "website": "Shopee.sg",
        "current_price": 321,
        "original_price": 349,
        "stock_status": "Limited Stock",
        "scraping_date": datetime.today().strftime('%Y-%m-%d')
    },
    {
        "product_name": "Galaxy Buds 2 Pro",
        "website": "Lazada.sg",
        "current_price": 199,
        "original_price": 299,
        "stock_status": "In Stock",
        "scraping_date": datetime.today().strftime('%Y-%m-%d')
    }
]

df = pd.DataFrame(data)

# Calculate discount %
df["discount_percent"] = ((df["original_price"] - df["current_price"]) / df["original_price"] * 100).round(1)

plt.figure(figsize=(8, 5))
plt.bar(df["product_name"], df["current_price"])
plt.title("Wireless Earbuds Price Comparison")
plt.xlabel("Product")
plt.ylabel("Current Price ($)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(df["product_name"], df["discount_percent"])
plt.title("Discount % on Wireless Earbuds")
plt.xlabel("Product")
plt.ylabel("Discount (%)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

stock_counts = df["stock_status"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(stock_counts, labels=stock_counts.index, autopct="%1.1f%%", startangle=140)
plt.title("Stock Availability of Wireless Earbuds")
plt.show()

# Simulated time-series data
trend_data = {
    "scraping_date": ["2025-10-20", "2025-10-21", "2025-10-22", "2025-10-23"],
    "Sony WF-1000XM4": [250, 240, 235, 228],
    "AirPods Pro": [349, 339, 325, 321],
}

trend_df = pd.DataFrame(trend_data)
trend_df["scraping_date"] = pd.to_datetime(trend_df["scraping_date"])
trend_df.set_index("scraping_date", inplace=True)

# Plot
trend_df.plot(figsize=(8, 5), marker='o')
plt.title("Price Trend of Wireless Earbuds Over Time")
plt.xlabel("Date")
plt.ylabel("Price ($)")
plt.grid(True)
plt.tight_layout()
plt.show()

```

The above code explained- first three lines of code is the data setup(pandas is used to organize data into a table (DataFrame).
matplotlib.pyplot is used to create charts.
datetime is used to record the current date (for scraping_date)and preparation. Then create sample data to represent on the graphs(used genAI for sample data), then created a price comparision bar chart plt.bar() plots product names on the x-axis and prices on the y-axis.
tight_layout() avoids label overlap.
show() displays the chart.; a discount % bar chart; a stock availaiblity pie chart- value_counts() counts how many products are in each stock status category.
plt.pie() creates a pie chart showing the percentage of:
In Stock,
Out of Stock,
Limited Stock
autopct="%1.1f%%" shows the percentage values on the chart. And a line graph for the price trends- Simulates daily price changes for products.
Converts the date column into a datetime index.
Plots each product’s price across time as a line graph.
Useful for tracking trends and seeing when discounts happen.
